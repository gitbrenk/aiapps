```python
prices = {
    "box_of_spaghetti" : 4,
    "lasagna"  : 5,
    "hamburger" : 2
   }
quantity = {
    "box_of_spaghetti" : 6,
    "lasagna"  : 10,
    "hamburger" : 0
    }

money_spent = 0
for items in prices:
    if (prices[items] >= 5):
        money_spent = money_spent + prices[items] * quantity[items]
money_spent
```
