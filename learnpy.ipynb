```python
prices = {
    "box_of_spaghetti" : 4,
    "lasagna"  : 5,
    "hamburger" : 2
   }
quantity = {
    "box_of_spaghetti" : 6,
    "lasagna"  : 10,
    "hamburger" : 0
    }

money_spent = 0
for items in prices:
    if (prices[items] >= 5):
        money_spent = money_spent + prices[items] * quantity[items]
money_spent
```

```python
products_on_sale, sale_prices, quantities = ['Chair_Type_1', 'Chair_Type_2', 'Chair_Type_3', 'Chair_Type_4'], [100, 120, 135, 150], [1000, 1500, 1300]

sales_revenue = [[chair, price * quantity] for chair in products_on_sale for price in sale_prices for quantity in quantities]
print(sales_revenue)
```
